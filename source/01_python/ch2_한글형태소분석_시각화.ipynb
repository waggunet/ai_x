## 3.1 HanNanum

````python
from konlpy.tag import Hannanum
hannanum = Hannanum(jvmpath=None,
                   max_heap_size=1024) #  기본값이 1024
hannanum.analyze(text)
````

````python
```
- 구글에서 konlpy docs 검색 후 https://konlpy.org/
- api 메뉴에서 "Comparison between POS tagging classes" 검색 후 =>  "See also
Korean POS tags comparison chart 클릭
```
````

````python
# 형태소 분석 : morphs
print(hannanum.morphs(text))
````

````python
# 명사만 추출
print(hannanum.nouns(text))
````

````python
# 품사태그
print(hannanum.pos(text)) # 품사갯수 기본값 ntags=9
````

````python
print(hannanum.pos(text, ntags=22))
````
