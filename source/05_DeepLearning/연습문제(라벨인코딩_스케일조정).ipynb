# 2. 지역명의 라벨인코딩
- 지역명을 라벨인코딩한 지역명2 필드를 추가
- 분석할 경우 원핫인코딩까지 할 것을 추천


```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 데이터 불러오기
try:
    df = pd.read_csv('C:/ai_x/download/shareData/부동산_250213/최종전국평당분양가격(결측치제외).csv', encoding='utf-8')
except UnicodeDecodeError:
    df = pd.read_csv('C:/ai_x/download/shareData/부동산_250213/최종전국평당분양가격(결측치제외).csv', encoding='cp949')

# 지역명 라벨인코딩
le = LabelEncoder()
df['지역명2'] = le.fit_transform(df['지역명'])

print("라벨인코딩 후 데이터 샘플 ('지역명2' 추가):")
print(df.head())
```

```python
# 지역명 원핫인코딩
df_onehot = pd.get_dummies(df, columns=['지역명'])

# 원핫인코딩된 전체 데이터프레임을 CSV로 저장
df_onehot.to_csv('C:/ai_x/download/shareData/부동산_250213/최종전국평당분양가격(결측치제외)_원핫인코딩.csv', index=False)

print("원핫인코딩 후 데이터 샘플:")
print(df_onehot.head().iloc[:, -10:])
```

# 3. normalization 스케일 조정
- 입력변수(지역명2, 연도, 월)와 타겟변수(평당분양가격) 따로 스케일 조정(MinMaxScaler 이용)
- 지역명2n, 연도n, 월n필드 추가


```python
from sklearn.preprocessing import MinMaxScaler

# 입력 변수와 타겟 변수 선택
X = df[['지역명2', '연도', '월']]
y = df[['평당분양가격']]

# 입력 변수 스케일링
scaler_x = MinMaxScaler()
df[['지역명2n', '연도n', '월n']] = scaler_x.fit_transform(X)

# 타겟 변수 스케일링
scaler_y = MinMaxScaler()
df['평당분양가격n'] = scaler_y.fit_transform(y)

print("Min-Max 스케일 조정 후 데이터 샘플:")
print(df[['지역명2', '연도', '월', '평당분양가격', '지역명2n', '연도n', '월n', '평당분양가격n']].head())
```

# 4. standarization 스케일 조정
- 입력변수와 타겟변수 따로 스케일 조정(StandardScaler 이용)
- 지역명2s, 연도s, 월s필드 추가


```python
from sklearn.preprocessing import StandardScaler

# 입력 변수 스케일링
std_scaler_x = StandardScaler()
df[['지역명2s', '연도s', '월s']] = std_scaler_x.fit_transform(X)

# 타겟 변수 스케일링
std_scaler_y = StandardScaler()
df['평당분양가격s'] = std_scaler_y.fit_transform(y)

print("Standard 스케일 조정 후 데이터 샘플:")
print(df[['지역명2', '연도', '월', '평당분양가격', '지역명2s', '연도s', '월s', '평당분양가격s']].head())
```
